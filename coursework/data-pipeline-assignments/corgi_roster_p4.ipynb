```python
# 2

import pandas as pd

# thiscsv
thiscsv = 'fdderby.csv'
```

```python
# 3

# read in csv as df, skip first row, 'Place' index col, df to console
df = pd.read_csv(thiscsv, skiprows=1, index_col = 'Place')

print(df)
```

```python
# 4 

# round 'Weight' to whole num - change to int .astype()
df['Weight']= round(df['Weight']).astype(int)

print(df['Weight'])
```

```python
# 5 

# 'Breed' col -> 'Pembroke' to 'Pem' & 'Cardigan' to 'Cardi' | df to console
df.replace(to_replace = ['Pembroke', 'Cardigan'], value = ['Pem', 'Cardi'], inplace=True)

print(df)
```

```python
# 6

# str.split() to split Name (Age)
df[['Name', 'Age']] = df['Name (Age)'].str.split("(", expand=True)

# str.replace('old', 'new') remove ')'
df['Age'] = df['Age'].str.replace(')', '')

# make Age numerical val
df['Age'] = pd.to_numeric(df['Age'])


# remove Name (Age) and any other weird cols df to console - I am not going to remove time. I don't know if we need it or not yet
df.drop(columns=['Name (Age)'], inplace=True)
print(df)
```

```python
# 7

# replace dnf to -1:00.0
df['Time'].replace(to_replace= 'DNF', value='-1:00.0', inplace= True)

# split on the : for Min (on left) Sec (on right)
# have to make time column a string first because it's giving me errors
df['Time'] = df['Time'].astype(str)
df[['Min', 'Sec']] = df['Time'].str.split(':', expand = True)

# pd.to_numeric for min and sec cols
df['Min'] = pd.to_numeric(df['Min'])
df['Sec'] = pd.to_numeric(df['Sec'])

# concat new time col for total secs
df['Time'] = (df['Min']*60) + df['Sec']

# replace neg vals
df['Time'].replace(to_replace= -60.00, value=-1.00, inplace= True)

# df to console
print(df)
```

```python
# 8

# housecleaning
df.drop(columns=['Min', 'Sec'], inplace = True)

# rearrange order of col df to console
df = df[['Name', 'Breed', 'Gender', 'Age', 'Weight', 'Time']]
print(df)
```
